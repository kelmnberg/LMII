```python
#1
def pivot(A,b,j):
  l = len(A)
  index = j
  Max = abs(A[j][j])
  for i in range(j,l):
    if abs(A[i][j]) > Max :
      Max, index = abs(A[i][j]) , i    #Max = abs(A[i][j) and index = i
  if index != j :

    A[[index,j]]  =  A[[j,index]]
    b[[index,j]]  = b[[j,index]]#use numpy tables



#2
def elim_bas(A,b,j):
    l = len(A)
    pivot(A,b,j)
    for e in range(j+1,l):
      #pivot(A,b,j)
      factor = A[e][j]/A[j][j]

      A[e] = A[e]-A[j]*factor
      b[e] = b[e] - b[j] * factor




#3
def descente(A,b):
  c = len(A[0])
  for j in range(0,c):
    pivot(A,b,j)
    elim_bas(A,b,j)
```
